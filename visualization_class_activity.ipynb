Matplotlib and numpy installation

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

Reading the pokemon csv and storing it in a data variable

```python
df = pd.read_csv("sample_data/pokemon.csv")
```

Reading the data to ensure its reflecting as intended

```python
print(df.head())
```

Removing missing values

```python
clean_data = df.dropna()
```

Creation of a scatter plot

```python
clean_data.plot.scatter(x="Attack", y = "Defense")
plt.show()
```

Creation of a histogram

```python
clean_data["Speed"].hist()
plt.title("Speed Frequency Distribution")
plt.xlabel("Speed")
plt.ylabel("Frequency")
plt.show()
```

Creation of a box plot

```python
clean_data.boxplot()
plt.title("Pokemon Statistics")
plt.xlabel("Stats")
plt.ylabel("Values")
plt.tight_layout()
plt.show()
```

Creation of a subplot containing the three graphs

```python
# Blue print for the subplot
fig, axes = plt.subplots(1, 3, figsize=(15,5))

# Scatter plot
axes[0].scatter(clean_data["Attack"], clean_data["Defense"])
axes[0].set_title("Attack Vs Defense")
axes[0].set_xlabel("Attack")
axes[0].set_ylabel("Defense")

# Histogram
axes[1].hist(clean_data["Speed"])
axes[1].set_title("Speed Distribution")
axes[1].set_xlabel("Speed")
axes[1].set_ylabel("Frequency")

# Box plot
axes[2].bar(clean_data["Type 1"].value_counts().index,
            clean_data["Type 1"].value_counts().values)
axes[2].set_title("Pokemon by Type")
axes[2].set_xlabel("Type")
axes[2].set_ylabel("Count")
axes[2].tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```

Creation of a correlation heatmap using Seaborn

```python
# Calculation of correlation between every numerical column in the dataset
correlation = clean_data.corr(numeric_only=True)

# Conversion of the matrix into a visual heatmap
sns.heatmap(correlation, annot = True)

plt.tight_layout()
plt.show()
```
