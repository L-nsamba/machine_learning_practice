```python
import numpy as np

x = np.array([1, 2])
y = np.array([3, 6])

# Setting of w and b
b = np.array([3, 3]) # bias
w = np.array([1, 1]) # weight
```

Calculation of ŷ

```python
def predict(x): # predict x is like y hat in this case
  return np.matmul(w, x.T) + b # Equivalent of  ŷ = wx + b
print(f'ŷ is equal to {predict(x)}')
```

Calculation of the Mean Square Error

```python
def MSE(x, y):
  return np.mean((predict(x) - y)**2) # Equivalent to (summation of (ŷ - y)**2)/ n)
print(f'The MSE is equal to {MSE(x, y)}') # Where ŷ = wx + b
```

Calculation of the Root Mean Square Error

```python
def RMSE(x, y):
  return np.sqrt(MSE(x, y))
print(f'The RMSE Error is equal to {RMSE(x, y):.2f}')
```

Calculating the derivative of w and b

```python
def dC_dw(x, y):
  return 2 * np.mean(x * (predict(x) - y)) # (2 * summation of x * (ŷ - y)) / n
print(f'The derivative of the weight dC_dw is equal to {dC_dw(x, y)}')

def dC_db(x, y):
  return 2 * np.mean(predict(x) - y) # (2 * summation of x * (ŷ - y)) / n
print(f'The derivative of the bias dC_db is equal to {dC_dw(x, y)}')
```

```python
# Calculating w(new) and b(new)
def update_parameters(w, b, x, y, learning_rate):
  w = w - learning_rate * dC_dw(x, y)
  b = b - learning_rate * dC_db(x, y)
  return w, b

# Setting of w and b
w = np.array([1.0, 1.0])
b = np.array([3.0, 3.0])

learning_rate = 0.1
error = []

# Looping 10 times to record the error
for i in range(0, 11):
  print(f'The Error is equal to {MSE(x, y):.2f}')
  error.append(np.round(MSE(x, y),2))
  w, b = update_parameters(w, b, x, y, learning_rate)

error = np.array(error)
print(error)
```

Visualization of the loss curve with learning rate

```python
import matplotlib.pyplot as plt
iterations = np.arange(0, 11)
plt.plot(iterations, error)
plt.title('The Loss Curve (With Learning Rate)')
plt.xlabel('This axis shows Loops/steps')
plt.ylabel('This axis shows error for each step')
plt.show()
```
