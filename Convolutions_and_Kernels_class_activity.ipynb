```python
import cv2
from PIL import Image, ImageFilter
import matplotlib.pyplot as plt
```

```python
# 1. Load the original image
img = Image.open('/content/dog.jpg')

# 2. Resize the image (to 28x28)
resized_img = img.resize((28, 28)) # Resize, we are technically blurring it

# 3. Apply a convolution filter (Finding edges on a grayscale version)
gray_img = img.convert('L') # Converting an image to a grayscale
convolved_img = gray_img.filter(ImageFilter.FIND_EDGES)

# 4. Create a side-by-side plot layout
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Plot Original
axes[0].imshow(img)
axes[0].set_title(f"Original Image\nSize: {img.size}")
axes[0].axis('off')

# Plot Resized (28x28)
axes[1].imshow(resized_img)
axes[1].set_title("Resized Image\nSize: (28, 28)")
axes[1].axis('off')

# Plot Convolved (Edge Detection)
axes[2].imshow(convolved_img, cmap='gray')
axes[2].set_title("Convolved Image\n(Edge Detection)")
axes[2].axis('off')

# Render the output inside Colab
plt.tight_layout()
plt.show()
```
